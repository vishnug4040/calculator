```python
import tkinter as tk
from tkinter import messagebox
import math

def calculate():
    try:
        num1 = float(entry1.get())
        num2 = float(entry2.get())
        operation = operation_var.get()

        if operation == "Addition":
            result = num1 + num2

        elif operation == "Subtraction":
            result = num1 - num2

        elif operation == "Multiplication":
            result = num1 * num2

        elif operation == "Division":
            if num2 == 0:
                messagebox.showerror(
                    "Error",
                    "Cannot divide by zero!"
                )
                return
            result = num1 / num2

        elif operation == "Power":
            result = num1 ** num2

        elif operation == "Square Root":
            if num1 < 0:
                messagebox.showerror(
                    "Error",
                    "Cannot find square root of a negative number!"
                )
                return
            result = math.sqrt(num1)

        
        messagebox.showinfo(
            "Calculation Result",
            f"Result: {result}"
        )

    except ValueError:
        messagebox.showerror(
            "Invalid Input",
            "Please enter valid numbers!"
        )

def clear():
    entry1.delete(0, tk.END)
    entry2.delete(0, tk.END)
    operation_var.set("Addition")

def exit_app():
    window.destroy()

window = tk.Tk()
window.title("Scientific Calculator")
window.geometry("500x550")
window.configure(bg="#2f1e3a")
window.resizable(False, False)

title_label = tk.Label(
    window,
    text="SCIENTIFIC CALCULATOR",
    font=("Arial", 22, "bold"),
    bg="#2f1e3a",
    fg="#ffb400"
)

title_label.pack(pady=25)

main_frame = tk.Frame(
    window,
    bg="#402b3f",
    padx=30,
    pady=30
)

main_frame.pack(
    padx=30,
    pady=10,
    fill="both"
)

tk.Label(
    main_frame,
    text="Enter First Number",
    font=("Arial", 12, "bold"),
    bg="#402b3f",
    fg="#f4e9ff"
).pack(anchor="w", pady=(0, 5))

entry1 = tk.Entry(
    main_frame,
    font=("Arial", 14),
    width=30,
    justify="center",
    bg="#fff4e0",
    fg="#2f1e3a",
    relief="flat"
)

entry1.pack(pady=(0, 15), ipady=8)

tk.Label(
    main_frame,
    text="Enter Second Number",
    font=("Arial", 12, "bold"),
    bg="#402b3f",
    fg="#f4e9ff"
).pack(anchor="w", pady=(0, 5))

entry2 = tk.Entry(
    main_frame,
    font=("Arial", 14),
    width=30,
    justify="center",
    bg="#fff4e0",
    fg="#2f1e3a",
    relief="flat"
)

entry2.pack(pady=(0, 15), ipady=8)

tk.Label(
    main_frame,
    text="Select Operation",
    font=("Arial", 12, "bold"),
    bg="#402b3f",
    fg="#f4e9ff"
).pack(anchor="w", pady=(0, 5))


operation_var = tk.StringVar()
operation_var.set("Addition")

operations = [
    "Addition",
    "Subtraction",
    "Multiplication",
    "Division",
    "Power",
    "Square Root"
]


operation_menu = tk.OptionMenu(
    main_frame,
    operation_var,
    *operations
)

operation_menu.config(
    font=("Arial", 12),
    bg="#cc00a8",
    fg="white",
    activebackground="#aa0088",
    activeforeground="white",
    width=20,
    relief="flat"
)

operation_menu["menu"].config(
    font=("Arial", 11),
    bg="#fff4e0",
    fg="#2f1e3a"
)

operation_menu.pack(pady=15)

button_frame = tk.Frame(
    main_frame,
    bg="#402b3f"
)

button_frame.pack(pady=20)

calculate_button = tk.Button(
    button_frame,
    text="CALCULATE",
    font=("Arial", 12, "bold"),
    bg="#9b59b6",
    fg="white",
    activebackground="#8e44ad",
    activeforeground="white",
    width=14,
    height=2,
    relief="flat",
    cursor="hand2",
    command=calculate
)

calculate_button.grid(
    row=0,
    column=0,
    padx=5
)

clear_button = tk.Button(
    button_frame,
    text="CLEAR",
    font=("Arial", 12, "bold"),
    bg="#3498db",
    fg="white",
    activebackground="#2980b9",
    activeforeground="white",
    width=14,
    height=2,
    relief="flat",
    cursor="hand2",
    command=clear
)

clear_button.grid(
    row=0,
    column=1,
    padx=5
)

exit_button = tk.Button(
    main_frame,
    text="EXIT",
    font=("Arial", 12, "bold"),
    bg="#d63384",
    fg="white",
    activebackground="#a02463",
    activeforeground="white",
    width=30,
    height=2,
    relief="flat",
    cursor="hand2",
    command=exit_app
)

exit_button.pack(pady=10)

footer = tk.Label(
    window,
    text="Simple • Fast • Scientific",
    font=("Arial", 10, "italic"),
    bg="#2f1e3a",
    fg="#d9b8e8"
)

footer.pack(pady=10)

window.mainloop()
```
